```python
import pandas as pd
import seaborn as sns
titanic_df = pd.read_csv("../train.csv")
```

```python
info_count = titanic_df.Survived.value_counts()
info_count
```

```python
info_count.plot.bar(
    title = "Count of Survived Passenger",
    xlabel = "Survived (No = 0, Yes = 1)",
    ylabel = "No. of Passenger",
    yticks = list(range(0,600, 50)),
    legend = True,
    colormap = "coolwarm",
    figsize = (4,3) 
    
)
```

```python
titanic_df[["Sex", "Survived"]].value_counts()
```

```python
ax = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex,
    # normalize="index"
    
).plot.bar(
    title = "Count of Survived Passenger",
    xlabel = "Survived (No = 0, Yes = 1)",
    ylabel = "No. of Passenger",
    yticks = list(range(0,600, 50)),
    legend = True,
    colormap = "coolwarm",
    figsize = (4,3) 
)

#Change x-axis tick lables
ax.set_xticklabels(['No','Yes'], rotation = 0)
```

```python
sns.countplot(
    data = titanic_df,
    x = "Survived",
    
)

```

## 2. Pie Chart

```python
group_by_embarked = titanic_df.groupby(by = "Embarked")
counter = group_by_embarked.count()
counter
```

```python
counter.plot(
    kind = "pie",
    y = "Survived",
    colormap= "Set3",
    autopct = "%1.1f%%",
    startangle = 90,
    
    
)
```

```python
ax = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex,
    # normalize="index"
    
).plot(
    kind = "pie",
    subplots = True,
    title = "Count of Survived Passenger",
    legend = True,
    colormap = "coolwarm",
    figsize = (4,3) 
)

#Change x-axis tick lables
# ax.set_xticklabels(['No','Yes'], rotation = 0)
```

## Handling Null /NaN/Na values

1. Removing Null Values
2. Fill something (Imputation)
3. Null as New Feature (Featuring Engineering)

```python
# Removing Null Values

embarked_removed = titanic_df.dropna(
    subset = "Embarked", axis = 0,
    ignore_index= True
)
embarked_removed.head()
```

```python
embarked_removed.head()
```

```python
embarked_removed.tail()
```

```python
titanic_df.isna().sum()
```

```python
#Null value as feature
embarked_removed.loc[:,"Cabin"] = embarked_removed.Cabin.notna().astype(int)
embarked_removed.head()
```

```python
embarked_removed.isna().sum()
```

```python
#Fill Null values (Imputation)
print(titanic_df.Age.mean())
print(titanic_df.Age.median())
```

```python
titanic_df.Age.plot(kind ="hist", bins = 20)
```

```python
med_age = embarked_removed.Age.median()
```

```python
median_replace_df = embarked_removed.copy()
median_replace_df["Age"] = median_replace_df.Age.fillna(
    value = med_age, axis = 0, 
)
```

```python
median_replace_df.isna().sum()
```

```python
#Fill Null values (Imputation)
print(titanic_df.Age.mean())
print(titanic_df.Age.median())

median_replace_df.Age.plot(kind="hist", bins = 20)
```

```python
# Imputation of Age with gender

male_med = embarked_removed[embarked_removed.Sex == "male"].Age.median()
female_med = embarked_removed[embarked_removed.Sex == "female"].Age.median()

male_med, female_med
```

```python
group_obj = embarked_removed.groupby(by="Sex")

mead_age = group_obj.Age.transform("median")
```

```python
median_gender_df = embarked_removed.copy()
median_gender_df["Age"] = median_gender_df.Age.fillna(
    value = med_age, axis = 0, 
)
```

```python
#Fill Null values (Imputation)
print(median_gender_df.Age.mean())
print(median_gender_df.Age.median())

median_gender_df.Age.plot(kind="hist", bins = 20)
```

```python
# Imputation Age with  Initials
embarked_removed.inti.value_counts()
```

```python
embarked_removed.loc[:,"inti"] = embarked_removed.inti.replace(
    to_replace = ["Dr", "Rev", "Don","Sir", "Col", "Capt", "Major"], value = "Mr"
)
```

```python
embarked_removed.inti.replace(
    to_replace = "Dr", value = "Mr"
)
```

```python
embarked_removed.loc[:, "inti"] = embarked_removed.inti.replace(
    to_replace = {
        "Mlle" : "Miss",
        "Mme" : "Mrs",
        "Ms" : "Miss",
        "Lady" : "Miss",
        "the Countess" : "Miss",
        "Jonkheer" : "Master"
    }
)
```

```python
mead_age = embarked_removed.groupby(by = "inti").Age.transform("median")
mead_age
```

```python
median_inti_df = embarked_removed.copy()

median_inti_df["Age"] = median_inti_df.Age.fillna(
    value = med_age, axis = 0
)
```

```python
print(median_inti_df.Age.mean())
print(median_inti_df.Age.median())

median_inti_df.Age.plot(kind = "hist", bins = 20)
```
